```python
import math

nama = "Giant"
usia = 20
tinggi_badan = 1.75
status_olahraga = False
 
berat_str = "97.5"
berat_badan = float(berat_str)
print(f"Halo, saya {nama}. Usia saya {usia} tahun dan berat saya {berat_badan} kg")

bmi = berat_badan / (tinggi_badan**2)
print(f"BMI Asli: {bmi}")
print(f"BMI Dibulatkan (2 desimal): {round(bmi, 2)}")


target_air = 2000
ukuran_botol = 350

hasil_asli = target_air / ukuran_botol  # True division
botol_penuh = target_air // ukuran_botol  # Floor division
sisa_air = target_air % ukuran_botol  # Modulo

print(f"Target botol (asli): {hasil_asli}")
print(f"Botol penuh: {botol_penuh}")
print(f"Sisa air: {sisa_air} ml")


kalori = 2100.5
kalori_bawah = math.floor(kalori)
kalori_atas = math.ceil(kalori)

print(f"Kalori dibulatkan ke bawah: {kalori_bawah}")
print(f"Kalori dibulatkan ke atas: {kalori_atas}")
```
